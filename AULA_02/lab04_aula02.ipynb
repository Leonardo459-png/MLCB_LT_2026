```
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

# 1. Dataset próprio
dados = {
    'mensagem': [
        'Quero comprar uma passagem para São Paulo',
        'Preciso de uma passagem para o Rio',
        'Como faço para comprar uma passagem?',
        'Quero reservar um voo para amanhã',

        'Quero cancelar minha reserva',
        'Preciso cancelar minha passagem',
        'Como faço para cancelar o voo?',
        'Quero desistir da minha reserva',

        'Quero falar com um atendente',
        'Preciso de ajuda de uma pessoa',
        'Quero falar com alguém da agência',
        'Pode me passar para um atendente?'
    ],
    'intencao': [
        'comprar_passagem',
        'comprar_passagem',
        'comprar_passagem',
        'comprar_passagem',

        'cancelar_reserva',
        'cancelar_reserva',
        'cancelar_reserva',
        'cancelar_reserva',

        'falar_atendente',
        'falar_atendente',
        'falar_atendente',
        'falar_atendente'
    ]
}

df = pd.DataFrame(dados)

# 2. Separação dos dados
X = df['mensagem']
y = df['intencao']

# 3. Divisão em treino e teste
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# 4. Vetorização
vectorizer = CountVectorizer()
X_train_vec = vectorizer.fit_transform(X_train)

# 5. Treinamento do modelo
modelo = LogisticRegression()
modelo.fit(X_train_vec, y_train)

# 6. Frases inéditas
novas_frases = [
    'Gostaria de adquirir um voo para Recife',
    'Preciso cancelar o meu voo',
    'Quero conversar com um funcionário'
]

novas_frases_vec = vectorizer.transform(novas_frases)
predicoes = modelo.predict(novas_frases_vec)

# 7. Resultados
print('--- RESULTADOS DO LAB 04 ---')

for frase, intencao in zip(novas_frases, predicoes):
    print(f"Mensagem: '{frase}' ==> Intenção: [{intencao}]")
```
